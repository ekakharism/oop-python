Soal 1: Membuat Class dan Object
Buatlah sebuah class Vehicle yang memiliki atribut:
* merk

* tahun

Tambahkan metode info_vehicle() yang menampilkan informasi kendaraan (merk dan tahun).
Goals: Memahami konsep dasar class dan object, serta cara mengakses atribut dan metode dalam class.


```python
class Vehicle:
    def __init__(self,merk,tahun):
        self.merk = merk 
        self.tahun = tahun  

    def info_vehicle(self):
        print(f'Kendaraan ini merknya adalah {self.merk} dengan tahun produksi {self.tahun}')

kendaraan1 = Vehicle('Yamoho',2020)

```

```python
kendaraan1.info_vehicle()
```

Soal 2: Mengimplementasikan Encapsulation
Kembangkan class Vehicle dengan menambahkan atribut private:

__bahan_bakar (dalam liter)

Tambahkan metode:

isi_bahan_bakar(jumlah) → menambah bahan bakar.

pakai_bahan_bakar(jumlah) → mengurangi bahan bakar jika cukup.

cek_bahan_bakar() → menampilkan sisa bahan bakar.

Jika bahan bakar tidak mencukupi, tampilkan pesan: "Bahan bakar tidak mencukupi".

Goals: Memahami encapsulation dan penggunaan access modifiers dalam class Python.


```python
class Vehicle:
    def __init__(self,merk,tahun):
        self.merk = merk 
        self.tahun = tahun 
        self.__bahan_bakar = 0 

    def info_vehicle(self):
        print(f'Kendaraan ini merknya adalah {self.merk} dengan tahun produksi {self.tahun}')

    def isi_bahan_bakar(self,jumlah):
        self.__bahan_bakar += jumlah

    def pakai_bahan_bakar(self,jumlah):
        if jumlah >= self.__bahan_bakar:
            print('Bensin tidak cukup')
        else :
            self.__bahan_bakar -= jumlah 

    def cek_bahan_bakar(self):
        print(f'Sisa bahan bakar saat ini adalah {self.__bahan_bakar} liter')

kendaraan1 = Vehicle('Yamoho',2020)
```

```python
kendaraan1.isi_bahan_bakar(100)
```

```python
kendaraan1.cek_bahan_bakar()
```

```python
kendaraan1.pakai_bahan_bakar(299)
```

Mobil

atribut tambahan: jenis_mobil

Truk

Setiap subclass harus:

Mewarisi atribut dan metode dari Vehicle.
Memiliki konstruktor sendiri yang memanggil super().
Goals: Memahami inheritance dan bagaimana membangun relasi antara superclass dan subclass.


```python
class Motor(Vehicle):
    def __init__(self,merk,tahun,jenis_motor):
        super().__init__(merk,tahun)
        self.jenis_motor = jenis_motor
    def info_vehicle(self):
        print(f'Kendaraan ini merk {self.merk} dengan tahun produksi {self.tahun} tipe {self.jenis_motor}')

```

Soal 4: Polymorphism dan Method Overriding

Di class Vehicle, buat metode default:
jalan(jarak) yang menampilkan: "Kendaraan berjalan sejauh [jarak] km"

Override metode jalan() di setiap subclass dengan format berbeda:
Motor
"Motor [merk] jenis [jenis_motor] berjalan sejauh [jarak] km.
Mobil
"Mobil [merk] jenis [jenis_mobil] berjalan sejauh [jarak] km.
Truk
"Truk [merk] berjalan sejauh [jarak] km membawa muatan [kapasitas_muatan] ton"
Goals: Mengimplementasikan polymorphism dan method overriding agar satu metode dapat memiliki perilaku berbeda pada subclass.


```python
class Vehicle:
    def __init__(self,merk,tahun):
        self.merk = merk 
        self.tahun = tahun 
        self.__bahan_bakar = 0 

    def info_vehicle(self):
        print(f'Kendaraan ini merknya adalah {self.merk} dengan tahun produksi {self.tahun}')

    def isi_bahan_bakar(self,jumlah):
        self.__bahan_bakar += jumlah

    def pakai_bahan_bakar(self,jumlah):
        if jumlah >= self.__bahan_bakar:
            print('Bensin tidak cukup')
        else :
            self.__bahan_bakar -= jumlah 

    def cek_bahan_bakar(self):
        print(f'Sisa bahan bakar saat ini adalah {self.__bahan_bakar} liter')

    def jalan(self,jarak):
        print(f'Kendaraan berjalan sejauh {jarak} km')

kendaraan1 = Vehicle('Yamoho',2020)
```

```python
class Motor(Vehicle):
    def __init__(self,merk,tahun,jenis_motor):
        super().__init__(merk,tahun)
        self.jenis_motor = jenis_motor
    def info_vehicle(self):
        print(f'Kendaraan ini merk {self.merk} dengan tahun produksi {self.tahun} tipe {self.jenis_motor}')
    def jalan(self,jarak):
        print(f'Motor {self.merk} jenis {self.jenis_motor} berjalan sejauh {jarak} km')

motor1 = Motor('HANDA',2020,'Matic')
```

```python
class Mobil(Vehicle):
    def __init__(self,merk,tahun,jenis_mobil):
        super().__init__(merk,tahun)
        self.jenis_mobil = jenis_mobil
    def info_vehicle(self):
        print(f'Kendaraan ini merk {self.merk} dengan tahun produksi {self.tahun} tipe {self.jenis_mobil}')
    def jalan(self,jarak):
        print(f'Mobil {self.merk} jenis {self.jenis_mobil} berjalan sejauh {jarak} km')

mobil1 = Mobil('Nissin',2020,'EV')
```

```python
class Truk(Vehicle):
    def __init__(self,merk,tahun,kapasitas_muatan):
        super().__init__(merk,tahun)
        self.kapasitas_muatan = kapasitas_muatan
    def info_vehicle(self):
        print(f'Kendaraan ini merk {self.merk} dengan tahun produksi {self.tahun} tipe {self.kapasitas_muatan}')
    def jalan(self,jarak):
        print(f'Truk {self.merk} berjalan sejauh {jarak} km membawa muatan {self.kapasitas_muatan} ton ')

truk1 = Truk('Trukinas',2000,200)
```

```python
kendaraan1.jalan(50)
motor1.jalan(100)
mobil1.jalan(120)
```

```python
kendaraan = [
    Motor('HANDA',2020,'Matic'),
    Mobil('Nissin',2020,'EV'),
    Truk('Trukinas',2000,200)
]

for k in kendaraan:
    k.jalan(50)
```
